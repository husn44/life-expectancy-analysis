## Data Preprocessing

#### Handling Missing Values

```python
# check for missing values
print(df.isnull().sum())
```

```python
# remove the leading/trailing spaces in the dataset
df.columns = df.columns.str.strip()
```

```python
# drop rows where target or adult mortality is missing
df.dropna(subset=['Life expectancy', 'Adult Mortality'], inplace=True)
```

For Life Expectancy and Adult Mortality, I chose to remove the missing rows because these are the main variables we are analyzing and since they only represent 0.3% of the dataset, removing them ensures higher data integrity by avoiding the use of estimated values for our main objectives

```python
# list of all other columns that still have missing values
cols_to_fix = [
    'Alcohol', 'Hepatitis B', 'BMI', 'Polio', 'Total expenditure',
    'Diphtheria', 'GDP', 'Population', 'thinness  1-19 years',
    'thinness 5-9 years', 'Income composition of resources', 'Schooling'
]
```

```python
# apply the imputation
for col in cols_to_fix:
    # 1) fill with the mean of the Country
    df[col] = df.groupby('Country')[col].transform(lambda x: x.fillna(x.mean()) if x.notnull().any() else x)
    
    # 2) if still missing, fill with the Status (Developed/Developing) mean
    df[col] = df.groupby('Status')[col].transform(lambda x: x.fillna(x.mean()))
```

I performed Statistical Imputation using a hierarchical Mean Imputation approach. Instead of applying a single global average, I used a two-step logic to ensure the filled data remained contextually accurate

First, I filled missing values using the average of the same Country over different years. Since health and economic data for a specific nation usually follow a consistent trend, using the country's own historical average is the most accurate way to preserve the unique characteristics of that nation

For countries that were missing data for an entire variable across all years, I used the mean of their Status group (Developed or Developing). This ensures that a missing value in a Developing nation is filled with data from economically similar peers rather than a global average that might be skewed by wealthy nations

```python
# verify that no missing values remain
print("Missing values after cleaning:\n")
print(df.isnull().sum())
```

#### Handling Noise or Outliers

```python
cols_to_cap = [
    'Measles', 'infant deaths', 
    'under-five deaths', 'HIV/AIDS'
]
```

I only applied capping to selected variables because not all extreme values are considered problematic. For variables like Measles and infant deaths, the outliers are extremely high and can distort the analysis, so capping helps reduce their impact.

However, for variables such as GDP and Population, the large values represent real-world differences between countries, not anomalies. Therefore, capping them would remove meaningful information, so they were retained or transformed instead.

```python
plt.figure(figsize=(12, 10))

for i, col in enumerate(cols_to_cap, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(y=df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
cols_to_cap = [
    'Measles', 'infant deaths', 
    'under-five deaths', 'HIV/AIDS'
]

# the cleaning loop
for col in cols_to_cap:
    # calculate the Q1,Q3 and IQR
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    # calculate the upper limit and lower limit using 1.5x IQR rule
    upper_limit = Q3 + 1.5 * IQR
    lower_limit = Q1 - 1.5 * IQR
    
    # apply Winsorization (capping): 
    # values above the upper_limit are set to the upper_limit.
    # values below the lower_limit are set to the lower_limit
    df[col] = df[col].clip(lower=lower_limit, upper=upper_limit)
```

I used the IQR Method to handle outliers in scale-based variables. This code first calculates the statistical boundaries for each column. Then, instead of deleting the rows, it uses Winsorization (the .clip() function) to cap extreme values at the upper and lower limits. I used this method to reduced the statistical noise and prevented extreme values from biasing. This approach was chosen over row deletion to preserve the integrity of the dataset and maintain a consistent sample size across all variables

#### Data Transformation

```python
# encoding the 'Status' column
# Developing = 1, Developed = 0
df['Status'] = df['Status'].map({'Developing': 1, 'Developed': 0})
```

I first performed Encoding on the 'Status' column to convert its categorical labels, "Developing" and "Developed," into the numerical values 1 and 0. This step is necessary because mathematical models cannot process text directly and by transforming these labels into numbers, the model can now factor in a country's development status during its calculations.

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

# get all numeric columns
all_numeric = df.select_dtypes(include=['float64', 'int64']).columns.tolist()

# dont need to scale 'Year' because it's time
# don't need scale 'Status' because it's already 0 and 1
to_skip = ['Year', 'Status']

# create the final list of columns to scale
cols_to_scale = [c for c in all_numeric if c not in to_skip]

# apply min-max normalization
df[cols_to_scale] = scaler.fit_transform(df[cols_to_scale])
```

I applied Normalization using the Min-Max Scaling method to all continuous numerical variables. Normalization ensures that all data falls within a small specified range, in this case, between 0 and 1. This step is crucial because variables in this dataset have vastly different scales, for instance, GDP is measured in thousands while HIV/AIDS is measured in small decimals. Without normalization, the model might incorrectly assume that GDP is more important simply because its numbers are larger. By squashing all values into the same 0-1 range, I ensured that every feature is treated with equal weight, preventing scale-based bias and leading to a more accurate and balanced result.
